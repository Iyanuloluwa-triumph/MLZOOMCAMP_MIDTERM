```python
def categorize(score):
    if score < 200:
        return 0        # fail
    elif score <= 300:
        return 1        # average
    else:
        return 2        # excellent

df3['label'] = df3['JAMB_Score'].apply(categorize)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, ConfusionMatrixDisplay, accuracy_score, confusion_matrix


X = df3.drop(['JAMB_Score', 'label'], axis=1)
y = df3['label']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=22)

clf = RandomForestClassifier()
clf.fit(X_train, y_train)
```

```python
preds = clf.predict(X_test)
print(classification_report(y_test, preds))
print("Accuracy:", accuracy_score(y_test, preds))
```

```python
import matplotlib.pyplot as plt
custom_labels = ['Low', 'Medium', 'High']

cm = confusion_matrix(y_test, preds)


# 4. Plot using ConfusionMatrixDisplay, supplying the custom labels
fig, ax = plt.subplots(figsize=(7, 6))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm, 
    display_labels=custom_labels # <-- This replaces 0, 1, 2 on the axes
)
disp.plot(cmap=plt.cm.Blues, values_format='d', ax=ax)

# Add relevant titles and axis labels
ax.set_title('Confusion Matrix with Custom Labels')
ax.set_xlabel('Predicted Label')
ax.set_ylabel('True Label')

plt.show()
```

And all of this must be inside one single pipeline object so your web app can call:

prediction = pipeline.predict(user_inputs)


This is exactly what scikit-learn Pipelines + Custom Transformers were designed for.

Let me show you the correct way to do it.

✅ STEP 1 — Build a Custom Transformer

This is where you compute your engineered features.

Example: Suppose you calculate:

study_efficiency = study_time * teacher_quality

parental_score = parent_education_level / 10

or any other engineered column

You wrap this logic inside a custom transformer:

from sklearn.base import BaseEstimator, TransformerMixin

class FeatureEngineer(BaseEstimator, TransformerMixin):
    def __init__(self):
        pass

    def fit(self, X, y=None):
        return self  # nothing to fit

    def transform(self, X):
        X = X.copy()

        # Example engineered features
        X["study_efficiency"] = X["study_time"] * X["teacher_quality"]
        X["parental_score"] = X["parent_education"] / 10

        return X


This step is crucial:
It makes your feature engineering reproducible, consistent, and web-deployable.

✅ STEP 2 — Add a ColumnTransformer for Scaling/Encoding

Example:

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_cols = ["study_time", "teacher_quality", "parent_education",
                "study_efficiency", "parental_score"]

categorical_cols = ["gender", "school_type"]

preprocess = ColumnTransformer([
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(), categorical_cols)
])


The key is:

The engineered columns must appear here
because the pipeline will pass transformed data step-by-step.

✅ STEP 3 — Build the Full Pipeline Including Your Model
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

pipeline = Pipeline([
    ("features", FeatureEngineer()),        # custom engineered features
    ("preprocess", preprocess),             # scaling & encoding
    ("model", LogisticRegression())         # your classifier
])


This single object now includes EVERYTHING:

feature engineering

new column creation

scaling

encoding

classification

✅ STEP 4 — Train the Pipeline Once
pipeline.fit(X_train, y_train)


All internal steps run automatically.
